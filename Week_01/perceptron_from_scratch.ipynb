```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
class Perceptron:
  def __init__(self,eta,epochs):
    self.weights = np.random.randn(3) * 1e-4
    print(f"Self.weights {self.weights}")

    self.eta = eta
    self.epochs = epochs


  def activationFunction(self,input,weights):
    z = np.dot(input,weights)
    return np.where(z>0,1,0)

  def fit(self,X,y):
    self.X = X
    self.y = y

    for epoch in range(self.epochs):
      print("Epoch: ",epoch)
      x_with_bias = np.c_[self.X,-np.ones((len(self.X),1))]
      print(f"X with bias {x_with_bias}")
      y_pred = self.activationFunction(x_with_bias,self.weights)
      print(f"Predicted value {y_pred}")

      error = self.y - y_pred
      print(f"Error \n{error}")
      #weight update

      self.weights = self.weights + self.eta * np.dot(x_with_bias.T,error)
      print(f"Updated weight \n {self.weights}")


def predict(self,X):
  x_with_bias = np.c_[X,-np.ones((len(X),1))]

  return self.activationFunction(x_with_bias,self.weights)


data = {
    "x1" : [0,0,1,1],
    "x2" : [0,1,0,1],
    "y" : [0,0,0,1]
}
AND = pd.DataFrame(data)

X = AND.drop('y',axis=1)
y = AND['y']
```

```python
#Model Prediction
model = Perceptron(eta = 0.5,epochs=10)
model.fit(X,y)
```
